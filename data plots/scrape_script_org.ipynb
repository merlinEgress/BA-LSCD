```python
def setConn(url):
    print("Entering: ", url)
    retries = 0
    backoff = BACKOFF_BASE

    while True:
        try:
            driver.get(url)
            time.sleep(BASE_DELAY + random.random() * JITTER)

            page_source = driver.page_source
            page_soup = BeautifulSoup(page_source, 'html.parser')
            return page_soup
        
        except (TimeoutException, WebDriverException) as e:
            if retries < MAX_RETRIES:
                wait = min(backoff, BACKOFF_MAX)
                print(f"Error fetching {url}: {e}. Retrying in {wait}s...")
                time.sleep(wait + random.random())
                retries += 1
                backoff *= 2
            else:
                print(f"Giving up on {url} after {retries} retries.")
                return None
```

```python
# splitting the work on two browsers/workers to speed up the process
# save for future use
import threading
import time
import random
import queue
from urllib.parse import urlparse
import urllib.robotparser
from selenium import webdriver
from selenium.webdriver.chrome.options import Options as ChromeOptions
from selenium.common.exceptions import WebDriverException, TimeoutException
from selenium.webdriver.common.by import By

#store scraped results thread-safely
from queue import Queue
results_queue = Queue()
results_queue_pages = Queue()

# -----------------------
# CONFIG: tweak these
# -----------------------
NUM_BROWSERS = 2               # total number of worker browsers (we'll spawn Chrome + Safari)
BROWSER_TYPES = ["chrome", "chrome"]  # order matters: worker0 -> chrome, worker1 -> safari
# If you're not on macOS, set both to "chrome"
PER_HOST_MIN_INTERVAL = 2.0    # base delay between requests *to the same host* (seconds)
JITTER = 1.0                   # add up to this many seconds of random jitter
PER_HOST_CONCURRENCY = 1       # how many concurrent requests allowed per host
MAX_RETRIES = 2
BACKOFF_BASE = 3.0             # seconds (doubled each retry)
BACKOFF_MAX = 120.0
USER_AGENT = "MyCrawler/1.0 (+https://example.com/bot-info; contact@example.com)"

# detection thresholds
MIN_PAGE_LENGTH_FOR_OK = 500   # if page source < this and no error shown, suspicious
CONSECUTIVE_ERROR_THRESHOLD = 5  # after this many errors, slow globally
GLOBAL_SLOWDOWN_MULTIPLIER = 2.5

# -----------------------
# State + synchronization
# -----------------------
url_q = queue.Queue()
stop_event = threading.Event()

# per-host semaphores and last-request timestamps
host_locks = {}
host_last_request = {}
host_lock = threading.Lock()  # protects host_locks & host_last_request

# global error tracker
error_lock = threading.Lock()
consecutive_errors = 0

# -----------------------
# Helpers
# -----------------------
def get_host(url):
    return urlparse(url).netloc

def get_host_semaphore(host):
    with host_lock:
        if host not in host_locks:
            host_locks[host] = threading.Semaphore(PER_HOST_CONCURRENCY)
            host_last_request[host] = 0.0
        return host_locks[host]

def polite_wait_for_host(host, slowdown_mult=1.0):
    """Ensure at least PER_HOST_MIN_INTERVAL (with jitter) since last request to host."""
    with host_lock:
        last = host_last_request.get(host, 0.0)
        now = time.time()
        interval = PER_HOST_MIN_INTERVAL * slowdown_mult
        # jitter applied each wait
        wait = max(0.0, interval - (now - last))
        if wait > 0:
            time.sleep(wait + random.random() * JITTER)
        # update last request timestamp to now (caller will issue request)
        host_last_request[host] = time.time()

def allowed_by_robots(url):
    parsed = urlparse(url)
    robots_url = f"{parsed.scheme}://{parsed.netloc}/robots.txt"
    rp = urllib.robotparser.RobotFileParser()
    try:
        rp.set_url(robots_url)
        rp.read()
        return rp.can_fetch(USER_AGENT, url)
    except Exception:
        # conservative: if robots.txt cannot be fetched, allow but still be very polite
        return True

def create_driver(browser_type):
    """Create and return a Selenium WebDriver for the requested browser_type."""
    if browser_type.lower() == "chrome":
        options = ChromeOptions()
        # newer headless flag for modern Chrome; adjust if your Chrome version differs
        options.add_argument("--headless=True")
        options.add_argument("--disable-gpu")
        options.add_argument("--no-sandbox")
        options.add_argument("--window-size=1920,1080")
        options.add_argument(f"user-agent={USER_AGENT}")
        # avoid some automation flags that can flag the bot
        options.add_argument("--disable-blink-features=AutomationControlled")
        driver = webdriver.Chrome(options=options)
        # optional: set page load timeout
        driver.set_page_load_timeout(60)
        return driver
    elif browser_type.lower() == "safari":
        # Safari: ensure safaridriver is enabled on macOS.
        # Safari does not accept many capability options like Chrome; UA changes are non-trivial.
        driver = webdriver.Safari()
        driver.set_page_load_timeout(60)
        return driver
    else:
        raise ValueError("Unsupported browser_type: " + browser_type)

def blocked_detection(driver):
    """Heuristic checks for CAPTCHAs / blocks / empty pages.
       Returns True if we suspect the request was blocked."""
    try:
        title = ""
        try:
            title = driver.title or ""
        except Exception:
            title = ""
        page = driver.page_source or ""
        # common heuristics: 'captcha', 'too many requests', 'access denied', tiny page
        lower = (title + " " + page).lower()
        if any(k in lower for k in ("captcha", "human", "verify", "robots", "access denied", "too many requests", "blocked")):
            return True
        if len(page) < MIN_PAGE_LENGTH_FOR_OK:
            return True
        return False
    except Exception:
        return True

def record_error(delta=1):
    global consecutive_errors
    with error_lock:
        consecutive_errors += delta

def clear_errors():
    global consecutive_errors
    with error_lock:
        consecutive_errors = 0

# -----------------------
# Worker / fetch logic
# -----------------------
def worker_thread(worker_id, browser_type):
    global consecutive_errors
    driver = None
    slowdown_mult = 1.0

    try:
        driver = create_driver(browser_type)
        print(f"[W{worker_id}] Started {browser_type} driver.")
    except Exception as e:
        print(f"[W{worker_id}] Failed to start {browser_type} driver: {e}")
        return

    while not stop_event.is_set():
        try:
            url = url_q.get(timeout=2)
        except queue.Empty:
            # no more work (or keep waiting)
            # break if queue is empty and we want to stop; otherwise continue
            # here we continue waiting. If you want exit-on-empty, check a sentinel.
            continue

        host = get_host(url)

        # robots.txt check
        if not allowed_by_robots(url):
            print(f"[W{worker_id}] Skipping (robots): {url}")
            url_q.task_done()
            continue

        sem = get_host_semaphore(host)

        # acquire per-host semaphore
        with sem:
            # polite wait for host-level throttling (applies slowdown multiplier if many errors)
            with error_lock:
                if consecutive_errors >= CONSECUTIVE_ERROR_THRESHOLD:
                    slowdown_mult = GLOBAL_SLOWDOWN_MULTIPLIER
                else:
                    slowdown_mult = 1.0

            polite_wait_for_host(host, slowdown_mult=slowdown_mult)

            # fetch with retries + exponential backoff
            attempt = 0
            backoff = BACKOFF_BASE
            success = False
            while attempt <= MAX_RETRIES and not success:
                try:
                    print(f"[W{worker_id}] GET {url} (attempt {attempt+1})")
                    driver.get(url)
                    time.sleep(0.5 + random.random() * 0.8)  # small post-load pause to allow JS to run
                    # check for block heuristics
                    if blocked_detection(driver):
                        raise RuntimeError("Blocked or suspicious page content detected")
                    # success: do your parsing here
                    page_source = driver.page_source
                    #########scraping integration#############
                    output_page = getFullText(page_source)
                    
                    #########scraping integration#############
                    results_queue.put(output_page)  # or append to a shared list safely
                    print(f"[W{worker_id}] Scraped {output_page['title']}")

                    print(f"[W{worker_id}] Fetched OK: {url} (len={len(page_source)})")
                    success = True
                    clear_errors()
                  # or store/process page_source as needed
                except TimeoutException as e:
                    print(f"[W{worker_id}] Timeout for {url}: {e}")
                    record_error(1)
                except WebDriverException as e:
                    print(f"[W{worker_id}] WebDriver error for {url}: {e}")
                    record_error(1)
                except RuntimeError as e:
                    # blocked or suspicious
                    print(f"[W{worker_id}] Blocked suspicion for {url}: {e}")
                    record_error(1)
                except Exception as e:
                    print(f"[W{worker_id}] Other error for {url}: {e}")
                    record_error(1)

                if not success:
                    attempt += 1
                    if attempt <= MAX_RETRIES:
                        wait = min(backoff, BACKOFF_MAX) + random.random()
                        print(f"[W{worker_id}] Backing off {wait:.1f}s before retry (attempt {attempt+1})")
                        time.sleep(wait)
                        backoff *= 2
                    else:
                        print(f"[W{worker_id}] Giving up on {url} after {attempt} attempts")
            # end retry loop

        # end with sem (releases automatically)
        url_q.task_done()

        # small global jitter to avoid synchronized hits across workers
        time.sleep(0.2 + random.random() * 0.5)

    # cleanup
    try:
        driver.quit()
    except Exception:
        pass
    print(f"[W{worker_id}] Exiting.")

# -----------------------
# Startup & usage example
# -----------------------
def start_pool_and_crawl(urls):
    # fill queue
    for u in urls:
        url_q.put(u)

    # spawn workers
    threads = []
    for i in range(NUM_BROWSERS):
        browser_type = BROWSER_TYPES[i] if i < len(BROWSER_TYPES) else "chrome"
        t = threading.Thread(target=worker_thread, args=(i, browser_type), daemon=True)
        t.start()
        threads.append(t)

    # wait until queue is processed
    try:
        url_q.join()
    except KeyboardInterrupt:
        print("Interrupted by user, signalling stop...")
        stop_event.set()

    # give threads a moment to exit
    stop_event.set()
    for t in threads:
        t.join(timeout=5)
    
    results = []
    while not results_queue.empty():
        results.append(results_queue.get())

    print(f"Scraped {len(results)} articles")
    return pd.DataFrame(results)


```
